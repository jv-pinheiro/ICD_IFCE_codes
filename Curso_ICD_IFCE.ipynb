```python
# Importar as bibliotecas necessárias
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

# Definir os valores observados na tabela
observed_table = np.array([
    [100, 150],  # Homem
    [900, 850]   # Mulher
])

# Calcular o teste qui-quadrado
chi2_result, p_value, dof, expected_table = chi2_contingency(observed_table, correction=False)

# Criar DataFrames para exibir as tabelas observadas e esperadas
observed_df = pd.DataFrame(
    observed_table,
    columns=["Homem", "Mulher"],
    index=["Usa", "Não Usa"]
)

expected_df = pd.DataFrame(
    expected_table,
    columns=["Homem", "Mulher"],
    index=["Usa", "Não Usa"]
)

# Calcular o coeficiente de contingência
total_students = observed_table.sum()
contingency_coefficient = np.sqrt(chi2_result / (chi2_result + total_students))

# Exibir as tabelas e resultados
print("Tabela Observada:")
display(observed_df)

print("\nTabela Esperada:")
display(expected_df)

print(f"\nQui-quadrado calculado (χ²): {chi2_result:.4f}")
print(f"p-valor: {p_value:.4f}")
print(f"Graus de liberdade: {dof}")
print(f"Coeficiente de Contingência: {contingency_coefficient:.4f}")

# Interpretar o resultado
if p_value < 0.05:
    print("\nInterpretação:\nComo o p-valor é menor que 0,05, rejeitamos a hipótese nula. "
          "Isso indica que existe evidência estatística de dependência entre 'Sexo' e 'Uso'.")
else:
    print("\nInterpretação:\nComo o p-valor é maior que 0,05, não rejeitamos a hipótese nula. "
          "Isso indica que não há evidência estatística de dependência entre 'Sexo' e 'Uso'.")
```

```python
# Importar as bibliotecas necessárias
import numpy as np
import pandas as pd

# Dados da tabela corrigidos
dados = {
    "Agente": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "X": [2, 2.5, 2.9, 3.3, 4.1, 4.3, 7, 13],  # Todos os 10 valores
    "Y": [17.5, 18.5, 19.5, 22.2, 26.5, 26.6, 36.6, 38.4]  # Todos os 10 valores
}

# Criar DataFrame
df = pd.DataFrame(dados)

# Calcular as médias
x_bar = df["X"].mean()
y_bar = df["Y"].mean()

# Calcular os desvios padrão
std_x = df["X"].std(ddof=0)  # ddof=0 para população
std_y = df["Y"].std(ddof=0)         # ddof=0 para população

# Adicionar colunas para os cálculos intermediários
df["x - x̄"] = df["X"] - x_bar
df["y - ȳ"] = df["Y"] - y_bar
df["(x - x̄) / σ(x)"] = df["x - x̄"] / std_x
df["(y - ȳ) / σ(y)"] = df["y - ȳ"] / std_y
df["z_x * z_y"] = df["(x - x̄) / σ(x)"] * df["(y - ȳ) / σ(y)"]

# Calcular o coeficiente de correlação linear
r = df["z_x * z_y"].sum() / len(df)

# Mostrar a tabela e o coeficiente de correlação
from IPython.display import display  # Para exibir no Colab
print("Tabela de Cálculos Intermediários:")
display(df)

print(f"\nCoeficiente de Correlação Linear (r): {r:.4f}")
```
